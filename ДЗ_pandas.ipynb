# Завдання 4
Яка загальна виручка у продавця Luis?

```python
total_sales_luis = df.query('salesman == "Luis"')['sale_amt'].sum()
print(f"Загальна виручка у продавця Luis: {total_sales_luis}")
```

# Завдання 5

Скільки і якого товару було продано?

```python
total_units_per_item = df.groupby('item')['units'].sum()
display(total_units_per_item)
```
